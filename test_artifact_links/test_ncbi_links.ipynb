# NCBI artifact links: test examples

Shows what `artifact_links/ncbi_links.py` returns on a few typical cases.
Only small pages are read (the assembly report and the folder listing); no genome file is downloaded.

```python
import sys
sys.path.append("..")

import json
from artifact_links.ncbi_links import ncbi_links, LinkError


def show(accession):
    """Print the full accession and the links, or the problem if there is one."""
    try:
        result = ncbi_links(accession)
    except LinkError as error:
        print("PROBLEM:", error)
        return
    print("asked:", accession, "-> got:", result["accession"])
    print(json.dumps(result["links"], indent=2))
```

## 1. A RefSeq assembly (cat)

RefSeq assemblies are annotated by NCBI: every file should be there.

```python
show("GCF_018350175.1")
```

## 2. The GenBank version of the same assembly

Same genome, but not annotated by NCBI: the annotation files (GFF, GTF, CDS, RNA, proteins, annotation report) are `null`.

```python
show("GCA_018350175.1")
```

## 3. Accession without a version number (dog)

NCBI picks the latest version (`.5`); the result says which one was used.

```python
show("GCF_000002285")
```

## 4. An older version (dog, CanFam3.1)

Older versions are still on NCBI and still have their files.

```python
show("GCF_000002285.3")
```

## 5. Assembly name already known

When the name comes from the search results, NCBI's API is not asked again: only the folder is read.


```python
result = ncbi_links("GCF_018350175.1", "F.catus_Fca126_mat1.0")
print(result["accession"])
print(json.dumps(result["links"], indent=2))
```

## 6. Problems

Each case gives a readable message instead of crashing.

```python
for accession in ["GCF_999999999.1", "hello", "GWHBJIX01000000"]:
    show(accession)
    try:
        ncbi_links("GCF_018350175", "F.catus_Fca126_mat1.0")
    except LinkError as error:
        print("PROBLEM:", error)
```
