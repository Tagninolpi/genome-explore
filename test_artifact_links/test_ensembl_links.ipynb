# Ensembl artifact links: test examples

Shows what `artifact_links/ensembl_links.py` returns on a few typical cases.
Only small pages are read (Ensembl's search answer and the folder listings); no genome file is downloaded.
Ensembl sorts each genome's files into sub-folders (`genome/`, `geneset/`) with the same generic names for every species.

```python
import sys
sys.path.append("..")

import json
from artifact_links.ensembl_links import ensembl_links
from artifact_links.common import LinkError


def show(accession, folder_url=None):
    """Print the full accession and the links, or the problem if there is one."""
    try:
        result = ensembl_links(accession, folder_url)
    except LinkError as error:
        print("PROBLEM:", error)
        return
    print("asked:", accession, "-> got:", result["accession"])
    print(json.dumps(result["links"], indent=2))
    print("species:", result["species"])
```

## 1. A genome annotated by Ensembl (cat)

The same assembly as NCBI's GenBank cat (`GCA_018350175.1`), annotated by Ensembl itself.
`cds`, `feature_table` and the three reports are `null`: Ensembl never provides them.

```python
show("GCA_018350175.1")
```

## 2. Accession without its version number

Ensembl gives back the full accession (`.1`).

```python
show("GCA_018350175")
```

## 3. A RefSeq genome in Ensembl (Argentine ant)

For some genomes Ensembl reuses NCBI's RefSeq annotation instead of making its own: the accession starts with `GCF_`
and the folder path says `refseq/` instead of `ensembl/`.

```python
show("GCF_040581485.1")
```

## 4. One assembly, two annotations (spiny mouse)

`GCA_964271855.1` has one annotation made by Ensembl and one made by another team, each in its own folder.
Without a folder given, Ensembl's own annotation is taken; the other one can be chosen by giving its folder (`ftp_url`).

```python
show("GCA_964271855.1")
print()
show("GCA_964271855.1", "http://ftp.ebi.ac.uk/pub/ensemblorganisms/GCA/964/271/855/1/community/2025_08/")
```

## 5. Problems

Each case gives a readable message instead of crashing. `GCF_018350175.1` (the RefSeq cat) exists at NCBI but not in Ensembl.

```python
for accession in ["GCF_018350175.1", "GWHBHRQ00000000", "hello"]:
    show(accession)
```
