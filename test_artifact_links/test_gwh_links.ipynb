# GWH artifact links: test examples

Shows what `artifact_links/gwh_links.py` returns on a few typical cases.
Only small pages are read (GWH's search answer and the folder listing); no genome file is downloaded.
Only genomes submitted directly to GWH (accessions starting with `GWH`) are handled; GWH's copies of NCBI genomes are taken from NCBI.

```python
import sys
sys.path.append("..")

import json
from artifact_links.gwh_links import gwh_links
from artifact_links.common import LinkError


def show(accession, folder_url=None):
    """Print the full accession and the links, or the problem if there is one."""
    try:
        result = gwh_links(accession, folder_url)
    except LinkError as error:
        print("PROBLEM:", error)
        return
    print("asked:", accession, "-> got:", result["accession"])
    print(json.dumps(result["links"], indent=2))
```

## 1. An annotated genome (Yunnan snub-nosed monkey)

Every file GWH can have is there. `gtf` and the three reports are `null`: GWH never provides them.

```python
show("GWHBHRQ00000000")
```

## 2. A genome without annotation (dog)

GWH's public API gives links for the GFF, RNA and proteins of this genome, but they point to the folder: the files do not exist.
Reading the real folder shows only the genome sequence and the md5 file.

```python
show("GWHBJIX01000000")
```

## 3. Accession without its version number (giant grouper, 2025)

Recent GWH accessions have a version (`.1`). GWH's search gives back the full accession, which is the one used in the file names.
Older accessions have no version, and none is added to them (see 1 and 2).

```python
show("GWHKEQZ00000000")
```

## 4. Folder already known

When the folder comes from the search results (`ftp_directory`), GWH's search is not asked again: only the folder is read.

```python
show("GWHBHRQ00000000",
     "https://download.cncb.ac.cn/gwh/Animals/Rhinopithecus_bieti_R.bieti-PacBio-HiC-GenomeAssembly_GWHBHRQ00000000/")
```

## 5. Problems

Each case gives a readable message instead of crashing.

```python
for accession in ["GWHZZZZ99999999", "GCA_018350175.1", "hello"]:
    show(accession)
```
